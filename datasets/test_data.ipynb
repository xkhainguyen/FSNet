```python
filepath2 = "/home/anonymous/FSNet/datasets/nonsmooth_nonconvex/socp/random2025_socp_dataset_var100_ineq50_eq50_ex10000"
with open(filepath2, 'rb') as f:
    dataset2 = pickle.load(f)

print(dataset2.keys())
```

```python
print(dataset2['solve_time_sec'][0:5], dataset2['iter_count'][0], dataset2['obj_value'][0], dataset2['eq_l2'][0], dataset2['ineq_l2'][0])
```

```python
# compute total metrics: obj_mean, obj_max, eq_vio, eq_vio_l2_mean, eq_vio_l2_max,ineq_vio_l2_mean, ineq_vio_l2_max, sucess_rate, total seconds, total iterations
obj_values = np.array(dataset2['obj_value'])
eq_l2 = np.array(dataset2['eq_l2'])
ineq_l2 = np.array(dataset2['ineq_l2'])
solve_time_sec = np.array(dataset2['solve_time_sec'])
iter_count = np.array(dataset2['iter_count'])
n_samples = len(obj_values)
obj_mean = np.mean(obj_values)
obj_max = np.max(obj_values)
eq_vio_l2_mean = np.mean(eq_l2)
eq_vio_l2_max = np.max(eq_l2)
ineq_vio_l2_mean = np.mean(ineq_l2)
ineq_vio_l2_max = np.max(ineq_l2)
success = np.array(dataset2['success'])
success_rate = np.sum(success) / n_samples
total_seconds = np.sum(solve_time_sec)
iters_mean = np.mean(iter_count) 
print(f"obj_mean: {obj_mean:.2f}, obj_max: {obj_max:.2f}, eq_vio_l2_mean: {eq_vio_l2_mean:.2f}, eq_vio_l2_max: {eq_vio_l2_max:.2f}, ineq_vio_l2_mean: {ineq_vio_l2_mean:.2f}, ineq_vio_l2_max: {ineq_vio_l2_max:.2f}, success_rate: {success_rate:.2f}, total_seconds: {total_seconds:.2f}, iters_mean: {iters_mean:.2f}")
merit = 1 * obj_mean + 1e5 * eq_vio_l2_mean + 1e5 * ineq_vio_l2_mean
print(f"merit: {merit:.2f}")
```
