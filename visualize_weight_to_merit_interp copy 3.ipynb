```python
import os
import numpy as np
import matplotlib.pyplot as plt

SKIP = 1 # skip_first
LINE_WIDTH = 3
BASE_SIZE = 24
MERIT_SCALE = 1e4
# Professional colors
COLORS = {
    "vanilla": "#C97D77",     # muted brick red
    "pretraining": "#3E6E9E", # steel blue
    "ours": "#49C157",        # sage green (updated)
}

def plot_merit_interpolation(
    t,
    merit,
    *,
    title="best label warmstart",
    xlabel="alpha  (0 = best, 1 = final)",
    ylabel="merit ($\times 10^4$)",
    save_path=None,
    dpi=300,
    show=True,
    close=True,
):
    fig, ax = plt.subplots(figsize=(6, 6), constrained_layout=True, dpi=dpi)

    A_idx = int((0.5 + 0.0) / 2 * 200)
    B_idx = int((0.5 + 1.0) / 2 * 200)
    merit_scaled = merit / MERIT_SCALE

    ax.plot(t, merit_scaled, linewidth=LINE_WIDTH * 1.5, color=COLORS['ours'], linestyle='-')
    ax.scatter([0.0], [(merit_scaled[A_idx])], s=BASE_SIZE*10, marker='o', color='black')
    ax.scatter([1.0], [(merit_scaled[B_idx])], s=BASE_SIZE*20, marker='*', color='black')

    ax.set_title(title, fontsize=BASE_SIZE * 1.1, pad=10)
    # ax.set_xlabel(xlabel, fontsize=BASE_SIZE, fontweight='normal')
    ax.set_xlim(-0.25, 1.25)
    # ax.set_ylabel(ylabel, fontsize=BASE_SIZE, fontweight='normal')
    ax.set_ylim(1, 1e5)
    ax.set_yscale("log")
    ax.set_yticks([1, 10, 100, 1000, 10000, 100000])
    ax.set_yticklabels([r"$10^0$", r"$10^1$", r"$10^2$", r"$10^3$", r"$10^4$", r"$10^5$"])
    ax.text(
        0.02,
        0.98,
        r"$\times 10^4$",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=BASE_SIZE,
    )
    
    ax.minorticks_off()
    ax.locator_params(axis='x', nbins=4)
    ax.tick_params(axis='both', labelsize=BASE_SIZE, width=LINE_WIDTH, length=8, color='black')

    # ax.grid(True, alpha=0.3)

    # annotate endpoints
    ax.annotate(f"{(merit_scaled[A_idx]):.3g}", (t[A_idx], merit_scaled[A_idx]), textcoords="offset points", xytext=(-20, -30),fontsize=BASE_SIZE+1)
    ax.annotate(f"{(merit_scaled[B_idx]):.3g}", (t[B_idx], merit_scaled[B_idx]), textcoords="offset points", xytext=(-20, -30), fontsize=BASE_SIZE+1)

    for spine in ax.spines.values():
        spine.set_linewidth(LINE_WIDTH * 0.75)
        spine.set_color('black')

    # if save_path is not None:
    #     out_dir = os.path.dirname(save_path)
    #     if out_dir:
    #         os.makedirs(out_dir, exist_ok=True)
    #     fig.savefig(save_path, dpi=dpi, bbox_inches="tight")
    #     print(f"Saved plot to: {save_path}", flush=True)

    if show:
        plt.show()

    if close:
        plt.close(fig)


if __name__ == "__main__":
    # Load from the NPZ you saved earlier
    npz_path = "figures/merit_interpolation_penalty_best0.npz"  # EDIT if needed
    data = np.load(npz_path)

    t = data["t"]
    merit = data["merit"]

    plot_merit_interpolation(
        t,
        merit,
        show=True,   # set True if you want interactive display
    )
```
